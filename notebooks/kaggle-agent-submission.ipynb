# 💡 The Innovation

## The Problem

Feature engineering is often the most impactful yet tedious part of machine learning. Data scientists spend hours:
- Reading documentation to understand feature meanings
- Experimenting with transformations (log, square, ratios)
- Creating interaction features based on domain knowledge
- Testing each feature to see if it actually helps

**What if we could automate this entire process?**

## Our Solution: An Autonomous Feature Engineering Agent

We built an **LLM-powered agent** that:

| Innovation | Description |
|------------|-------------|
| 📄 **Reads Documentation** | Understands `data_description.txt` to learn what each column means |
| 📊 **Uses SHAP Feedback** | Analyzes which features matter most and targets improvements there |
| 🧪 **Self-Evaluates** | Tests each feature with 5-fold CV, keeps only what improves RMSLE |
| ⚡ **Runs in Parallel** | Producer-consumer pattern with batched Gemini API calls |
| 🧠 **Learns from Failure** | Remembers what didn't work to avoid repeating mistakes |

## 1. Imports and Setup

````python
import pandas as pd
import numpy as np
import lightgbm as lgb
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_log_error, mean_squared_error
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.preprocessing import StandardScaler
import xgboost as xgb
import warnings
warnings.filterwarnings('ignore')

print('✅ Imports successful!')
````

````python
def rmsle(y_true, y_pred):
    """Calculate Root Mean Squared Logarithmic Error"""
    return np.sqrt(mean_squared_log_error(y_true, y_pred))

def rmse(y_true, y_pred):
    """Calculate Root Mean Squared Error"""
    return np.sqrt(mean_squared_error(y_true, y_pred))

print('✅ Helper functions defined!')
````

````python
import os

# Environment detection: Kaggle or local
if os.path.exists('/kaggle/input'):
    DATA_PATH = '/kaggle/input/ai-assisted-hackathon-1'
    OUTPUT_PATH = '/kaggle/working'
    print('🏠 Running in Kaggle environment')
else:
    DATA_PATH = '../data' if os.path.exists('../data/train.csv') else 'data'
    OUTPUT_PATH = '../outputs/predictions' if os.path.exists('../outputs') else 'outputs/predictions'
    print('💻 Running in local environment')

os.makedirs(OUTPUT_PATH, exist_ok=True)

train = pd.read_csv(f'{DATA_PATH}/train.csv')
test = pd.read_csv(f'{DATA_PATH}/test.csv')

print(f'\n📊 Train shape: {train.shape}')
print(f'📊 Test shape: {test.shape}')
````

## 3. Preprocessing Class

````python
class MinimalPreprocessor:
    """
    True minimal preprocessing - agent discovers everything.
    
    Does ONLY:
    1. Numeric NaN → 0 (agent can detect absence via > 0)
    2. Categorical NaN → "None" string
    3. Convert categoricals to pd.Categorical dtype
    """
    
    def __init__(self):
        self.categories = {}
        self._fitted = False
    
    def fit(self, df):
        """Learn categories for each categorical column."""
        for col in df.select_dtypes(include=['object']).columns:
            unique_vals = list(df[col].dropna().unique())
            if 'None' not in unique_vals:
                unique_vals.append('None')
            self.categories[col] = sorted(set(unique_vals))
        self._fitted = True
        return self
    
    def transform(self, df):
        """Apply minimal transformation."""
        if not self._fitted:
            raise ValueError("Preprocessor must be fitted before transform")
        
        df = df.copy()
        
        # 1. Numeric NaN → 0
        for col in df.select_dtypes(include=['number']).columns:
            if col not in ['Id', 'SalePrice']:
                df[col] = df[col].fillna(0)
        
        # 2. Categorical NaN → "None", convert to Categorical dtype
        for col, cats in self.categories.items():
            if col in df.columns:
                df[col] = df[col].fillna('None')
                df[col] = pd.Categorical(df[col], categories=cats)
        
        return df
    
    def fit_transform(self, df):
        """Fit and transform in one step."""
        return self.fit(df).transform(df)


class OutlierRemover:
    """
    Remove known outliers from Ames Housing training data.
    
    The Ames Housing dataset has 2 well-known outliers:
    - Houses with GrLivArea > 4000 sq ft but SalePrice < $300,000
    These represent partial sales or recording errors.
    """
    
    def __init__(self, verbose=True):
        self.verbose = verbose
        self.removed_indices = []
    
    def remove_known_outliers(self, df):
        """Remove the 2 famous Ames Housing outliers."""
        if 'SalePrice' not in df.columns:
            if self.verbose:
                print("   Skipping outlier removal (no SalePrice - test data)")
            return df
        
        # Identify outliers: large houses with suspiciously low prices
        mask = (df['GrLivArea'] > 4000) & (df['SalePrice'] < 300000)
        self.removed_indices = df[mask].index.tolist()
        
        removed_count = mask.sum()
        if removed_count > 0 and self.verbose:
            print(f"   Removed {removed_count} outlier(s)")
        
        return df[~mask].reset_index(drop=True)


print('MinimalPreprocessor class defined!')
print('OutlierRemover class defined!')
````

## 4. Agent-Discovered Features

These features were **automatically discovered** by our LLM agent. Each feature includes comments explaining when and how it was found.

````python
def apply_agent_features(df):
    """
    Features discovered by the Autonomous Feature Engineering Agent.
    
    These features were generated by Gemini LLM, tested with 5-fold CV,
    and kept only because they improved RMSLE.
    """
    df = df.copy()
    
    # Feature 1: OverallQual_GrLivArea_Product
    # Strategy: Interaction
    # Discovered: Iteration 1
    # Improvement: +0.00118 RMSLE (best improvement!)
    # Reasoning: Quality × size interaction captures premium value of large high-quality homes
    df['OverallQual_GrLivArea_Product'] = df['OverallQual'] * df['GrLivArea']
    
    # Feature 2: TotalLivingSF
    # Strategy: Aggregation
    # Discovered: Iteration 2
    # Improvement: +0.00045 RMSLE
    # Reasoning: Combines all living space into single metric
    df['TotalLivingSF'] = df['GrLivArea'] + df['TotalBsmtSF'].fillna(0) + df['GarageArea'].fillna(0)
    
    # Feature 3: GrLivArea_Cubic
    # Strategy: Polynomial
    # Discovered: Iteration 8
    # Improvement: +0.00024 RMSLE
    # Reasoning: Captures non-linear relationship - premium grows exponentially with size
    df['GrLivArea_Cubic'] = df['GrLivArea'] ** 3
    
    # Feature 4: LivAreaRatioToLot
    # Strategy: Size Efficiency
    # Discovered: Iteration 11
    # Improvement: +0.00005 RMSLE
    # Reasoning: Living area to lot ratio indicates space efficiency/density
    df['LivAreaRatioToLot'] = df['GrLivArea'] / df['LotArea']
    
    return df

print('✅ Agent-discovered features defined!')
print('   - OverallQual_GrLivArea_Product (interaction)')
print('   - TotalLivingSF (aggregation)')
print('   - GrLivArea_Cubic (polynomial)')
print('   - LivAreaRatioToLot (ratio)')
````

## 5. Preprocessing Pipeline

````python
# Extract target and log-transform
target = train['SalePrice']
train_ids = train['Id']
test_ids = test['Id']

# Step 1: Apply agent features FIRST (on raw data)
print('1. Applying agent-discovered features...')
train_with_agent = apply_agent_features(train)
test_with_agent = apply_agent_features(test)

# Step 2: Remove outliers from training data
print('\n2. Removing outliers...')
outlier_remover = OutlierRemover()
train_clean = outlier_remover.remove_known_outliers(train_with_agent)

# Update target after outlier removal
target = train_clean['SalePrice']
target_log = np.log1p(target)

# Step 3: Minimal preprocessing
print('\n3. Preprocessing data...')
preprocessor = MinimalPreprocessor()

X_train_raw = train_clean.drop(['Id', 'SalePrice'], axis=1)
X_test_raw = test_with_agent.drop(['Id'], axis=1)

train_processed = preprocessor.fit_transform(X_train_raw)
test_processed = preprocessor.transform(X_test_raw)

# Step 4: Extract numeric features only (for model compatibility)
print('\n4. Extracting numeric features...')
X_train = train_processed.select_dtypes(include=['number'])
X_test = test_processed.select_dtypes(include=['number'])

# Step 5: Align columns (test may be missing some features)
common_cols = list(set(X_train.columns) & set(X_test.columns))
X_train = X_train[common_cols]
X_test = X_test[common_cols]

# Step 6: Handle NaN values
X_train = X_train.fillna(0)
X_test = X_test.fillna(0)

print(f'\nTraining data shape: {X_train.shape}')
print(f'Test data shape: {X_test.shape}')
print(f'Total features: {X_train.shape[1]}')
````
